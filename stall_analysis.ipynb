**Task 1 - calculating the average**

1.   using sum(daily_sales)/ len(daily_sales) to fing the average
2.   using max(n) and min() to fing the numbers with high and low values in the list  




```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,

2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

"Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
"Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
"Sukuma":  {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
"Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
"Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
#to find average
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

#maximum sales
max_sales = max(daily_sales)
max_sales_day = days[daily_sales.index(max_sales)]
best_week_num = 1 if max_sales_day in days[:7] else 2

#minimum sales
min_sales = min(daily_sales)
min_sales_day = days[daily_sales.index(min_sales)]
worst_week_num = 1 if min_sales_day in days[:7] else 2

#output
print(f"The average daily sales is KES {average_sales:.2f}")
print(f"The maximum daily sales was KES {max_sales} on {max_sales_day} in week {best_week_num}")
print(f"The minimum daily sales was KES {min_sales} on {min_sales_day} in week {worst_week_num}")
print(f"Total Sales: KES{total_sales}")
```

**Task 2- calculating growth percentage**

1.  using list slicing to separate the weeks [:7] and [7:]
2.  for tuple unpacking using W1-total and w2_total to automatically assigns the two elements by the function.
3. the growth formula((current-previous)//previous) * 100 to compute the percentage change



```python
def week_totals(sales):
  week_1_total = sum(sales[:7])
  week_2_total = sum(sales[7:])
  return week_1_total, week_2_total

week_1_total, week_2_total = week_totals(daily_sales)
growth = ((week_2_total - week_1_total) / week_1_total) * 100

print(f"Week 1 Total: KES{week_1_total}")
print(f"Week 2 Total: KES{week_2_total}")
print(f"The growth percentage is {growth:.2f}%")
```

**Task 3- classigying good and bad days**

1.   using the Dictionary counting pattern to increase the tally dynamically for each category as it loops throuh the 14 days
2.   Classify_day function to compare the given amounts 20% above and below



```python
def classify_day(amount, average):
  if amount > average *1.20:
    return "strong"
  elif amount < average * 0.80:
    return "weak"
  else:
    return "normal"

category_counts = {"strong": 0, "weak": 0, "normal": 0}

for sale in daily_sales:
    category = classify_day(sale, average_sales)
    category_counts[category] += 1

print(f"Sales Category Breakdown:")
for category, count in category_counts.items():
  print(f"-{category}:{count} day(s)")
```

**Task 4- weekday vs weekend sales**

1.   using slicing both weeks into weekday and weekend slots
2.   using percentage difference to calculate the sales diff



```python
weekdays_sales = daily_sales[0:5] + daily_sales[7:12]
weekends_sales = daily_sales[5:7] + daily_sales[12:14]

#fo the average
average_weekday = sum(weekdays_sales) / len(weekdays_sales)
average_weekend = sum(weekends_sales) / len(weekends_sales)

pct_difference = ((average_weekend - average_weekday) / average_weekday) * 100

print(f"Average Weekday Sales: KES{average_weekday:.2f}")
print(f"Average Weekend Sales: KES{average_weekend:.2f}")
print(f"The percentage difference is {pct_difference:+.2f}%")
```

**Task 5- calculating the gross margin percentage**

1.   The <stock_dict.items()> unpacks both key ad its nested value i a single loop




```python
def check_reorder_and_margins(stock_dict):
  reorder_list = []

  print("--- Product Gross Margin Analysis ---")
  for product, details in stock_dict.items():
    qty = details["qty"]
    reorder = details["reorder"]
    cost = details["cost"]
    price = details["price"]

    gross_margin = (price - cost) / cost * 100
    print(f"{product}: {gross_margin:.2f}%")

    if qty < reorder:
      reorder_list.append(product)
  return reorder_list

items_to_reorder = check_reorder_and_margins(stock)

print("\nItems needing reorder(qty < reorder level):")
print(items_to_reorder)
```
